```python
# ==============================================================
# Standardize Property Addresses
# ==============================================================

# Property addresses are often stored in different formats.
#
# Examples:
#
# 123 Main St #2
# 123 MAIN STREET Apt 2
# 123 Main St.
#
# Standardizing addresses improves duplicate detection during the
# Data Cleaning stage.
```

```python
def clean_address(address):
    """
    Standardize property addresses.

    Processing steps
    ----------------
    1. Convert all text to uppercase.
    2. Remove apartment or unit numbers.
    3. Remove punctuation.
    4. Remove extra whitespace.
    5. Return the standardized address.
    """

    if pd.isna(address):
        return ""

    address = address.upper()

    # Remove apartment numbers.
    address = re.sub(r"#\w+", "", address)

    # Remove punctuation.
    address = re.sub(r"[^A-Z0-9 ]", "", address)

    # Remove extra spaces.
    address = re.sub(r"\s+", " ", address)

    return address.strip()
```

```python
# ==============================================================
# Create Standardized Address
# ==============================================================

# Apply the address cleaning function to every property record.

df_brooklyn["clean_address"] = (
    df_brooklyn["address"]
    .apply(clean_address)
)

df_queens["clean_address"] = (
    df_queens["address"]
    .apply(clean_address)
)
```

```python
# ==============================================================
# Standardize Column Names
# ==============================================================

# Zillow exports use column names that differ from the naming
# convention used throughout this project.
#
# Standardizing column names makes the datasets consistent and
# simplifies downstream preprocessing.
```

```python
df_queens = df_queens.rename(columns={
    "addresszipcode": "zip_code",
    "area": "living_area",
    "coordinates/latitude": "latitude",
    "coordinates/longitude": "longitude",
    "price": "listing_price",
    "rentzestimate": "rent_estimate",
    "zestimate": "zestimate",
    "hometype": "home_type",
    "daysonzillow": "days_on_zillow",
    "listingstatus": "listing_status",
    "taxassessedvalue": "tax_assessed_value"
})
```
